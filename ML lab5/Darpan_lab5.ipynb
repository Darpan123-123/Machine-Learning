**Prepare X (features) and y (price) for modeling: encode every categorical column into numeric form, and briefly 
justify your choice of encoding method for each type of column**

```python
y = df["SalePrice"]
X = df.drop(columns=["SalePrice", "Id"]) 

qual_scale = {"None": 0, "Po": 1, "Fa": 2, "TA": 3, "Gd": 4, "Ex": 5}

ordinal_maps = {
    "ExterQual": qual_scale, "ExterCond": qual_scale, "BsmtQual": qual_scale,
    "BsmtCond": qual_scale, "HeatingQC": qual_scale, "KitchenQual": qual_scale,
    "FireplaceQu": qual_scale, "GarageQual": qual_scale, "GarageCond": qual_scale,
    "PoolQC": qual_scale,
    "BsmtExposure": {"None": 0, "No": 1, "Mn": 2, "Av": 3, "Gd": 4},
    "BsmtFinType1": {"None": 0, "Unf": 1, "LwQ": 2, "Rec": 3, "BLQ": 4, "ALQ": 5, "GLQ": 6},
    "BsmtFinType2": {"None": 0, "Unf": 1, "LwQ": 2, "Rec": 3, "BLQ": 4, "ALQ": 5, "GLQ": 6},
    "GarageFinish": {"None": 0, "Unf": 1, "RFn": 2, "Fin": 3},
    "Functional": {"Sal": 0, "Sev": 1, "Maj2": 2, "Maj1": 3, "Mod": 4, "Min2": 5, "Min1": 6, "Typ": 7},
    "LandSlope": {"Sev": 0, "Mod": 1, "Gtl": 2},
    "LotShape": {"IR3": 0, "IR2": 1, "IR1": 2, "Reg": 3},
    "PavedDrive": {"N": 0, "P": 1, "Y": 2},
    "Fence": {"None": 0, "MnWw": 1, "GdWo": 2, "MnPrv": 3, "GdPrv": 4},
    "Utilities": {"ELO": 0, "NoSeWa": 1, "NoSewr": 2, "AllPub": 3},
}

# In columns mein NaN = "feature absent" hai, "None" se fill karo
fill_none_cols = ["BsmtQual", "BsmtCond", "BsmtExposure", "BsmtFinType1",
                   "BsmtFinType2", "FireplaceQu", "GarageFinish", "GarageQual",
                   "GarageCond", "PoolQC", "Fence"]
for col in fill_none_cols:
    X[col] = X[col].fillna("None")

for col, mapping in ordinal_maps.items():
    X[col] = X[col].map(mapping)

# -------------------------------------------------------------------
# 3. BINARY columns -> 0/1 (sirf 2 categories hain, ek column kaafi hai)
# -------------------------------------------------------------------

X["Street"] = X["Street"].map({"Grvl": 0, "Pave": 1})
X["CentralAir"] = X["CentralAir"].map({"N": 0, "Y": 1})

# -------------------------------------------------------------------
# 4. Baaki jo bhi non-numeric bacha, woh NOMINAL hai (koi natural order
#    nahi, jaise Neighborhood, RoofStyle) -> one-hot encode. Label-
#    encoding yahan galat hoti kyunki model wrongly ek fake ranking
#    seekh leta.
# -------------------------------------------------------------------
nominal_cols = X.select_dtypes(include=["object"]).columns.tolist()
print("One-hot encoded columns:", nominal_cols)
X = pd.get_dummies(X, columns=nominal_cols, drop_first=True)

# -------------------------------------------------------------------
# 5. Baaki numeric missing values (LotFrontage, MasVnrArea, GarageYrBlt)
#    median se fill karo — yeh encoding ka nahi, missing-data ka issue hai
# -------------------------------------------------------------------
num_na_cols = X.columns[X.isnull().any()].tolist()
print("Median-imputed numeric columns:", num_na_cols)
for col in num_na_cols:
    X[col] = X[col].fillna(X[col].median())

# -------------------------------------------------------------------
# 6. Confirm X fully numeric, no missing values (Task 5)
# -------------------------------------------------------------------
assert X.isnull().sum().sum() == 0
assert X.select_dtypes(include=["object"]).shape[1] == 0

print("\nFinal X shape:", X.shape)
print("Final y shape:", y.shape)
X.head()
```

**Task 13 In your own words, what does each metric tell you, and which one would you show to a non-technical 
stakeholder? Why?** 
## Ans 
**MAE (Mean Absolute Error) tells me the average size of the model's mistakes, in the same units as the price itself. On average, my predictions are off by about $21,926 — it treats every error equally, whether small or large.**

**MSE (Mean Squared Error) squares each error before averaging, which makes it punish large mistakes much more heavily than small ones. The downside is that its units are squared dollars, so the number itself (~2.57 billion) is hard to interpret directly — it's mainly useful for comparing models against each other.**

**RMSE (Root Mean Squared Error) is the square root of MSE, which brings it back into normal dollar units while still weighting big errors more heavily than MAE does. In my case, RMSE ($50,727) is noticeably higher than MAE ($21,926), which tells me a few large outlier errors (like the negative-price prediction seen in the residual plot) are pulling this number up.**

**R² (R-squared) tells me what fraction of the variation in house prices the model successfully explains. My R² of 0.66 means the model accounts for about 66% of the price variation, leaving 34% unexplained — a decent but not excellent fit.**

**14. Compute MAE, MSE, RMSE, and R² on the training set, and place them side by side with your Part 5 test-set

```python
y_train_pred =lr.predict(X_train)

# Training set metrics
mae_train = mean_absolute_error(y_train, y_train_pred)
mse_train = mean_squared_error(y_train, y_train_pred)
rmse_train = np.sqrt(mse_train)
r2_train = r2_score(y_train, y_train_pred)


comparison_metrics = pd.DataFrame({
    "Metric": ["MAE", "MSE", "RMSE", "R²"],
    "Training": [mae_train, mse_train, rmse_train, r2_train],
    "Testing": [mae, mse, rmse, r2]
})


comparison_metrics["Training"] = comparison_metrics["Training"].apply(lambda x: f"{x:,.2f}")
comparison_metrics["Testing"] = comparison_metrics["Testing"].apply(lambda x: f"{x:,.2f}")

print(comparison_metrics)
```

**Yes, there's a large gap — R² drops from 0.92 (training) to 0.66 (testing), and RMSE more than doubles from $21,656 to $50,727. This gap indicates overfitting: the model has memorized noise and quirks specific to the training data (likely due to having ~219 features but only ~1,168 rows) rather than learning patterns that generalize,
so it performs far worse on unseen houses.**
